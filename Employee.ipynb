```python
mean_age = df["Age"].mean()
print(mean_age)
```

```python
df["Age"] = df["Age"].fillna(mean_age)
df
```

```python
print(df["Age"].isnull().sum())
```

```python
print('Checking for null values after filling missing Age values:')
display(df.isnull().sum())
```

```python
mean_salary = df["Salary"].mean()
df["Salary"] = df["Salary"].fillna(mean_salary)

print('Missing values after filling Salary with its mean:')
display(df.isnull().sum())
```

```python
df
```

```python
df['Phone'] = df['Phone'].abs()
print("Converted 'Phone' column to absolute values. Displaying descriptive statistics for 'Phone' column to confirm:")
display(df['Phone'].describe())
```

```python
df
```

```python
df["Department_Region"].value_counts().plot(kind="bar")

plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Salary"], bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()
```

```python
department_salary = df.groupby("Department_Region")["Salary"].mean()

department_salary
```

```python
department_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)

plt.show()
```
